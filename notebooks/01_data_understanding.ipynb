```python
import pandas as pd
import numpy as np
import os
import zipfile
```

```python
zip_path = "../data/raw/student_resource.zip"

with zipfile.ZipFile(zip_path, "r") as zip_ref:
    zip_ref.extractall("../data/raw/extracted")

print("Dataset Extracted Successfully")
```

```python
for root, dirs, files in os.walk("../data/raw/extracted"):
    for file in files:
        print(os.path.join(root, file))
```

```python
csv_files = []

for root, dirs, files in os.walk("../data/raw/extracted"):
    for file in files:
        if file.endswith(".csv"):
            csv_files.append(os.path.join(root, file))

print("CSV Files Found:")
for file in csv_files:
    print(file)
```
